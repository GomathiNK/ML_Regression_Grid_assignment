```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=0,dtype=int)
```

```python
dataset
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children',  'sex_female', 'sex_male',
       'smoker_no', 'smoker_yes']]
```

```python
dependent=dataset[['charges']]
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid={'kernel':['rbf','sigmoid','linear'],'C':[10,100,1000,2000,3000],'gamma':['auto','scale']}
grid=GridSearchCV(SVR(),param_grid,refit=True, verbose=3,n_jobs=-1)
grid.fit(independent, dependent)
```

```python
re=grid.cv_results_
```

```python
print("R value {}:".format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
table
```

```python
# age_input=int(input("Age:")) 
bmi_input=float(input("BMI:"))
children_input=int(input("Children:"))
sex_male_input=int(input("Sex Male 0 or 1:"))
sex_female_input=int(input("Sex Female 0 or 1:"))
smoker_yes_input=int(input("Smoker Yes 0 or 1:"))
smoker_no_input=int(input("Smoker Yes 0 or 1:"))
```

```python
future_predict=grid.predict([[age_input,bmi_input,children_input,sex_male_input,sex_female_input,smoker_no_input,smoker_yes_input]])
print("Future_Prediction={}".format(future_predict))
```
